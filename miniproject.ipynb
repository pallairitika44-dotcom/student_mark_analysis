```python
#student marks analysis      #libraries
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

print("student marks analysis")
```

```python
data={
    'student':['ali','sara','ravi','john','priya'],
    'math':[85,92,45,78,60],
    'science':[78,88,50,82,70],
    'english':[90,95,40,70,65]
}

df=pd.DataFrame(data)
df.to_csv('student_marks.csv',index=False)
print("datasaved in csv")
print(df)
```

```python
df=pd.read_csv('student_marks.csv')

df['total']=df['math']+df['science']+df['english']
df['average']=df['total']/3
df['result']=df['average'].apply(lambda x:'pass'if x>50 else 'fail')

print("==== student marks analysis====")
print(df)
print()
print("topper:",df.loc[df['total'].idxmax(),'student'])
print("pass:",len(df[df['result']=='pass']))
print("fail:",len(df[df['result']=='fail']))

                     
```

```python
df = pd.read_csv('student_marks.csv')
df['Total'] = df['math'] + df['science'] + df['english']

plt.figure(figsize=(8, 5))
plt.bar(df['student'], df['Total'],
        color=['red', 'green', 'orange', 'blue', 'purple'])

plt.title('Student total Marks', fontsize=14)
plt.xlabel('Student Name', fontsize=12)
plt.ylabel('total Marks', fontsize=12)

for i, val in enumerate(df['Total']):
    plt.text(i, val + 1, str(val),
             ha='center', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
df=pd.read_csv('student_marks.csv')
df['Total']=df['math']+df['science']+df['english']
df['average']=df['Total']/3
df['result']=df['average'].apply(lambda x:'pass' if x>50 else 'fail')

result_counts=df['result'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(result_counts, labels=result_counts.index, 
        autopct='%1.1f%%', colors=['lightgreen', 'salmon'],shadow=True)
plt.title('pass vs fail distribution')
plt.show()

print("===final result===")
print(result_counts)


```
